```python
import pandas as pd
from pathlib import Path
import re

class FeatureEngineering:
    def __init__(self):
        # Path to the combined CSV file
        self.output_path: Path = Path.cwd().resolve().parent / "data" / "combined.csv"

        # Load dataset
        self.input_data: pd.DataFrame = pd.read_csv(self.output_path, encoding="cp950")

        # Apply feature engineering on initialization
        self.output_data: pd.DataFrame = self.apply_conversion()

    def apply_conversion(self):
        df = self.input_data.copy()

        # 🔹 One combined regex with named groups
        address_pattern = re.compile(
            r"(?P<district>.+?區)?"
            r"(?P<road>.+?(?:路|街|大道|橋|圓環|廣場))?"
            r"(?P<section>\d+)段?"
            r"(?P<lane>\d+)巷?"
            r"(?P<alley>\d+)(?:弄|衖)?"
            r"(?P<number>\d+)號?"
            r"(?P<floor>\d+)樓?"
            r"(?P<room>\d+)室?",
        )

        # Extract all components in one vectorized operation
        extracted = df["Location of accident"].str.extract(address_pattern)

        # Merge back into original DataFrame
        df = pd.concat([df, extracted], axis=1)

        # Convert numeric fields where applicable
        numeric_fields = ["section", "lane", "alley", "number", "floor", "room"]
        df[numeric_fields] = df[numeric_fields].apply(pd.to_numeric, errors="coerce")

        return df
```

```python
import re

def parse_taipei_address(address: str) -> dict:
    """
    Parse a Taipei address into its components.

    This function extracts common elements from a Taipei address, such as:
        - district (e.g., 大同區)
        - road (e.g., 重慶北路)
        - section (e.g., 3段)
        - lane (e.g., 137巷)
        - alley (e.g., 20弄 or 20衖)
        - building number (e.g., 5號)
        - floor (e.g., 3樓)
        - room (e.g., 101室)

    It returns a dictionary with these keys. Only the first match for each
    component is extracted, and once a component is found, it is removed
    from the remaining string to avoid overlapping matches.

    Parameters:
        address (str): A string containing a Taipei address.

    Returns:
        dict: A dictionary with keys:
            'district', 'road', 'section', 'lane', 'alley', 'number', 'floor', 'room'.
            If a component is not found, it will be absent from the dictionary.
    """

    # Ordered list of address components and their regex patterns
    structures = [
        ("district", r".+?區"),                      # Matches district ending with '區'
        ("road", r".+?(路|街|大道|橋|圓環|廣場)"),   # Matches road names with common suffixes
        ("section", r"\d+段"),                      # Matches section numbers, e.g., '3段'
        ("lane", r"\d+巷"),                         # Matches lane numbers, e.g., '137巷'
        ("alley", r"\d+弄|\d+衖"),                  # Matches alleys, e.g., '20弄' or '20衖'
        ("number", r"\d+號"),                        # Matches building numbers, e.g., '5號'
        ("floor", r"\d+樓"),                         # Matches floor numbers, e.g., '3樓'
        ("room", r"\d+室"),                          # Matches room numbers, e.g., '101室'
    ]

    result = {}           # Dictionary to store parsed components
    remaining = address   # Copy of address to remove matched parts progressively

    for key, pattern in structures:
        match = re.search(pattern, remaining)  # Search for the pattern in the remaining string
        if match:
            result[key] = match.group()        # Save the matched string
            # Remove matched part to prevent it from being matched again in other patterns
            remaining = remaining.replace(match.group(), "", 1)

    return result
```

```python
import numpy as np

def calc_distance_to_center(df, lat_col='CoordinateY', lon_col='CoordinateX'):
    """
    Adds a 'dist_to_center' column with distance in meters to Taipei city center.

    Parameters:
        df : pandas DataFrame
        lat_col : str, name of latitude column
        lon_col : str, name of longitude column

    Returns:
        df : DataFrame with 'dist_to_center' column
    """
    df = df.copy()  # avoid modifying original

    # City center coordinates
    lat0, lon0 = 25.0330, 121.5654

    # Convert degrees to radians
    lat = np.radians(df[lat_col].values)
    lon = np.radians(df[lon_col].values)
    lat0_rad = np.radians(lat0)
    lon0_rad = np.radians(lon0)

    # Haversine formula (vectorized)
    dlat = lat - lat0_rad
    dlon = lon - lon0_rad
    a = np.sin(dlat / 2) ** 2 + np.cos(lat0_rad) * np.cos(lat) * np.sin(dlon / 2) ** 2
    c = 2 * np.arcsin(np.sqrt(a))
    R = 6371000  # Earth radius in meters
    df['dist_to_center'] = R * c

    return df
```

```python
df = calc_distance_to_center(df)
df.head()
```

```python
# Shows how accidents are distributed by distance from city center; good for scatter/bar plots.
df_distance_accidents = df.groupby('dist_to_center').size().reset_index(name = "number_of_accidents")
df_distance_accidents.head()
```
