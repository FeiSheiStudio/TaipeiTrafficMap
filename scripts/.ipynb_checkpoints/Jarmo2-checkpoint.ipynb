```python
def haversine(lat1, lon1, lat2, lon2):
    """
    Compute Haversine distance in kilometers between two points.
    """
    lat1_rad, lon1_rad = np.radians(lat1), np.radians(lon1)
    lat2_rad, lon2_rad = np.radians(lat2), np.radians(lon2)

    delta_lat = lat2_rad - lat1_rad
    delta_lon = lon2_rad - lon1_rad

    a = np.sin(delta_lat/2)**2 + np.cos(lat1_rad) * np.cos(lat2_rad) * np.sin(delta_lon/2)**2
    c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1 - a))
    R = 6371.0  # Earth radius in km
    return R * c
```

```python
import polars as pl
import numpy as np

def haversine(lat1, lon1, lat2, lon2):
    """
    Compute Haversine distance in kilometers between two points.
    """
    lat1_rad, lon1_rad = np.radians(lat1), np.radians(lon1)
    lat2_rad, lon2_rad = np.radians(lat2), np.radians(lon2)

    delta_lat = lat2_rad - lat1_rad
    delta_lon = lon2_rad - lon1_rad

    a = np.sin(delta_lat/2)**2 + np.cos(lat1_rad) * np.cos(lat2_rad) * np.sin(delta_lon/2)**2
    c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1 - a))
    R = 6371.0  # Earth radius in km
    return R * c


def assign_nearest_mrt(accidents: pl.DataFrame, mrts: pl.DataFrame) -> pl.DataFrame:
    """
    Assign nearest MRT station info to accidents with Haversine distance.
    Adds: station_name, mrt_lon, mrt_lat, distance (km), Close_or_Not (1 if < 0.5 km), StationID, mrt_address.
    """
    # Collect coordinates and info as NumPy arrays
    acc_xy = accidents.lazy().select(["lon", "lat"]).collect().to_numpy()
    mrt_xy = mrts.lazy().select(["lon", "lat"]).collect().to_numpy()
    mrt_names = mrts.lazy().select("StationName").collect().to_numpy().ravel()
    mrt_ids = mrts.lazy().select("StationID").collect().to_numpy().ravel()
    mrt_address = mrts.lazy().select("StationAddress").collect().to_numpy().ravel()

    # Prepare arrays to store results
    nearest_station_name = []
    nearest_mrt_lon = []
    nearest_mrt_lat = []
    nearest_distance = []
    nearest_mrt_id = []
    nearest_mrt_address = []
    nearest_distance_within_500 = []

    # Compute nearest station for each accident
    for lat1, lon1 in acc_xy:  # acc_xy[:, 0] = lon, [:,1] = lat
        distances_km = haversine(lat1, lon1, mrt_xy[:,1], mrt_xy[:,0])  # lat/lon order
        idx = np.argmin(distances_km)

        nearest_station_name.append(mrt_names[idx])
        nearest_mrt_lat.append(mrt_xy[idx,1])  # lat
        nearest_mrt_lon.append(mrt_xy[idx,0])  # lon
        nearest_distance.append(distances_km[idx])
        nearest_distance_within_500.append(1 if distances_km[idx] < 0.5 else 0)
        nearest_mrt_id.append(mrt_ids[idx])
        nearest_mrt_address.append(mrt_address[idx])


    # Build result Polars DataFrame
    result = accidents.lazy().with_columns([
        pl.Series("station_name", nearest_station_name),
        pl.Series("mrt_lat", nearest_mrt_lat),
        pl.Series("mrt_lon", nearest_mrt_lon),
        pl.Series("distance_km", nearest_distance),
        pl.Series("Close_or_Not", nearest_distance_within_500),
        pl.Series("StationID", nearest_mrt_id),
        pl.Series("mrt_address", nearest_mrt_address)
    ]).collect()

    return result
```

```python
class FeatureEngineering:
    def __init__(self):
        self.agent = None
        self.mrts   = None
        self.accidents = None

    def initialize_agent(self):
        self.agent = DataImport()
        return self

    # Decorator must be defined as a normal method
    def files_exist(method):
        @wraps(method)
        def wrapper(self, *args, **kwargs):
            self.initialize_agent()
            if not (os.path.exists(self.agent.output_path_mrt)) or not (os.path.exists(self.agent.output_path_traffic)):
                print("Either MRT file is not available, or accident data is not processed!")
            return method(self, *args, **kwargs)
        return wrapper

    def run_agent(self):
        if not self.agent:
            self.initialize_agent()
            self.agent.save_all()

    @files_exist
    def read_data(self):
        self.mrts = pl.scan_csv(self.agent.output_path_mrt)
        self.accidents = pl.scan_csv(self.agent.output_path_traffic,
                                     schema_overrides = {"Time": pl.Datetime})
        return

    def preprocess_accidents(self):
        self.read_data()

        # Apply all processes
        self.accidents = self.accidents.pipe(self.collect_time).pipe(self.collect_address)
        return self

    def collect_time(self, lazy) -> pl.LazyFrame:
        date_params = ['year', 'month', 'day', 'hour', 'minute','second', 'time', 'weekday']
        lazy = lazy.with_columns(
            [
                getattr(pl.col('Time').dt, i)().alias(i) for i in date_params
            ]
        )
        lazy = lazy.with_columns(

            [
            # For classification, we assign accidents that are after half hour to the next whole hour
            pl.when(pl.col("minute") <= 30)
            .then(pl.col("hour"))
            .otherwise(pl.col('hour') + 1)
            .alias("new_hour")
            ]
        )
        return lazy

    def collect_address(self, lazy) -> pl.LazyFrame:
        patterns = self.address_pattern()
        return lazy.with_columns(
            [
                pl.col("Location of accident").str.extract(pattern, group_index=0).alias(col_name)
                for col_name, pattern in patterns
            ]
        )


    @staticmethod
    def address_pattern():
        return [
    ("district", r".+?區"),
    ("road", r"[^與\(]+?(路|街|大道|橋|圓環|廣場)"),
    ("section", r"\d+段"),
    ("lane", r"\d+巷"),
    ("alley", r"\d+弄|\d+衖"),
    ("number", r"\d+號"),
    ("floor", r"\d+樓"),
    ("room", r"\d+室")
        ]


```
